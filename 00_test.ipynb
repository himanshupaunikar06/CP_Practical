```python
#Create a simple if else else code to check mark and give
#grade A , B, C, D, F based on the mark entered by user.
mark = int(input('Enter your mark: '))
if 91 <= mark <= 100:
    print('Grade A')
elif 81 <= mark <= 90:
    print('Grade B')
elif 71 <= mark <= 80:
    print('Grade C')
elif 61 <= mark <= 70:
    print('Grade D')
else:
    print('Grade F')
```
